## 第三部分：CPython 如何實作這些型別？

除了知道 Python 使用哪些型別表示數學物件之外，

我們也希望了解：

> Python 物件在 CPython 中是如何實作的？

請查閱資料，完成下表。

| 數學概念 | Python 型別 | CPython 實作 | 主要特點 |
|----------|------------|-------------|----------|
| 自然數 | `int` | PyLongObject | 與整數共用 int，可表示任意精度整數 |
| 整數 | `int` | PyLongObject | 任意精度，可依數值大小增加所需儲存空間 |
| 實數 | `float` | PyFloatObject | 底層以 C double 表示，是有限精度的浮點數 |
| 複數 | `complex` | PyComplexObject | 由實部與虛部組成，其數值部分使用兩個 double |
| 集合 | `set` | PySetObject | 儲存不重複元素，小集合可使用物件內部空間，較大的集合使用額外的可變大小記憶體區塊 |
| 命題真假 | `bool` | PyBool_Type，Py_True / Py_False | bool 是 int 的子類別，而且只有 True、False 兩個布林物件 |
| 串列 | `list` | PyListObject | 可按照順序保存多個 Python 物件，長度可以改變 |

人工智慧模型需要儲存大量參數（Parameters）與權重（Weights）。

3. 什麼是 FP32？FP32 是 32 位元的單精度浮點數，具有不錯的精度與數值範圍，是常見的浮點格式
4. 什麼是 FP16？FP16 是 16 位元的半精度浮點數，占用的記憶體較少、運算效率較高，但精度與數值範圍比 FP32 小
5. 什麼是 BF16？BF16 是 16 位元的浮點數格式，精度比 FP32 低，但能保留較大的數值範圍，因此常用於 AI 模型的訓練與運算
6. 為什麼大型 AI 模型不直接使用 Python 的 `int` 作為權重？因為 AI 模型的權重通常需要表示正負小數，而 Python 的 int 只能表示整數，所以通常使用浮點數，而不是直接使用 int

### 思考題一

Python 官方文件指出：

```python
int
```

具有 Unlimited Precision（無限制精度）。

請回答：

1. Unlimited Precision 是什麼意思？Unlimited Precision 是指 Python 的 int 沒有固定的位元數限制，數字變大時可以動態使用更多記憶體來儲存，因此可以表示非常大的整數
2. 它是否代表可以儲存無限大的整數？不是，它代表沒有預先設定的固定整數上限，但實際能儲存多大的整數仍受到電腦可用記憶體和運算資源限制
3. CPython 如何實現這項特性？CPython 使用 PyLongObject 實作 Python 的 int，整數需要的空間會隨數值大小增加，因此不像固定寬度整數只能使用固定的位元數

### 思考題二

比較：

```python
int
```

與

```python
float
```

回答：

1. 為什麼 `int` 幾乎沒有固定上限？Python 的 int 採用任意精度設計，數字越大就配置更多記憶體來儲存，因此沒有固定的 32-bit 或 64-bit 整數上限，實際限制主要是電腦的記憶體與運算資源
2. 為什麼 `float` 卻存在範圍限制？Python 的 int 採用任意精度設計，數字越大就配置更多記憶體來儲存，因此沒有固定的 32-bit 或 64-bit 整數上限，實際限制主要是電腦的記憶體與運算資源
3. 為什麼 `float` 可能產生誤差？因為 float 使用有限位元的二進位方式表示數值，像 0.1、0.2 等十進位小數無法被有限的二進位位數精確表示，只能儲存近似值，因此計算時可能產生微小誤差，例如 0.1 + 0.2 會得到約 0.30000000000000004

### 思考題三

比較下列兩種設計理念：

#### Python 的整數

```text
Arbitrary Precision
正確性優先
```

#### AI 模型常用的 FP16、BF16

```text
Limited Precision
效率優先
```

請說明：

1. 兩種設計理念的差異。Arbitrary Precision（任意精度）會依整數大小增加儲存空間，盡可能保持整數計算的精確性；Limited Precision（有限精度）則使用固定數量的位元表示數值，允許一定程度的近似誤差，以換取較高的速度和較低的記憶體需求
2. 各自的優點與缺點。Python int：優點是整數可以精確表示，而且不受固定 32-bit 或 64-bit 上限限制；缺點是數字越大，需要的記憶體與計算成本也可能越高
FP16、BF16：優點是每個數值只需要 16 bit，可以降低記憶體需求並提高大量數值運算的效率；缺點是精度有限，因此可能產生捨入誤差，也無法精確表示所有數值
3. 為什麼大型語言模型通常選擇較低精度的表示方式？大型語言模型具有大量權重，而且訓練和推論需要進行大量矩陣運算，因此使用 FP16、BF16 等較低精度格式，可以減少模型占用的記憶體、降低資料傳輸量並提高支援硬體上的運算效率；AI 通常可以接受一定程度的數值近似，因此不一定需要每個權重都使用最高精度
4. 如果所有 AI 模型都改用 FP64，可能會產生哪些問題？如果全部改用 FP64，每個數值需要 64 bit，相較 FP16、BF16 的 16 bit，單就數值儲存而言需要約 4 倍空間；這會增加記憶體容量與頻寬需求，模型訓練和推論的硬體成本也可能提高，而且許多 AI 工作並不需要 FP64 提供的高精度，因此可能造成資源浪費
