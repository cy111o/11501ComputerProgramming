問題：撰寫一Python程式，為使用者輸入年份（可能是西元或民國）找尋其相對應的十二生肖。十二生肖是以十二年為一週期，每一年分別以一種動物來表示猴（monkey）、雞（rooster）、狗(dog)、豬(pig)、鼠(rat)、牛(ox)、虎(tiger)、兔(rabbit)、龍(dragon)、蛇（snake）、馬(horse)，以及羊(sheep) 。

def get_zodiac():
    print("=== 十二生肖查詢程式 ===")
    
    # 十二生肖列表（依據指定的順序排列，餘數 0 對應猴年）
    zodiacs = [
        "猴 (Monkey)",
        "雞 (Rooster)",
        "狗 (Dog)",
        "豬 (Pig)",
        "鼠 (Rat)",
        "牛 (Ox)",
        "虎 (Tiger)",
        "兔 (Rabbit)",
        "龍 (Dragon)",
        "蛇 (Snake)",
        "馬 (Horse)",
        "羊 (Sheep)"
    ]

    print("請選擇輸入的年份類型：")
    print("1. 西元年份 (例如: 2026)")
    print("2. 民國年份 (例如: 115)")
    
    choice = input("請輸入選項 (1 或 2): ").strip()

    if choice == "1":
        try:
            gregorian_year = int(input("請輸入西元年份："))
        except ValueError:
            print("輸入錯誤，請輸入有效的數字！")
            return
    elif choice == "2":
        try:
            roc_year = int(input("請輸入民國年份："))
            # 民國轉西元
            gregorian_year = roc_year + 1911
        except ValueError:
            print("輸入錯誤，請輸入有效的數字！")
            return
    else:
        print("無效的選項，請重新執行程式。")
        return

    # 計算生肖索引（以 2004 年猴年為基準，2004 % 12 = 0）
    index = gregorian_year % 12
    animal = zodiacs[index]

    # 輸出結果
    print("-" * 30)
    print(f"西元年份：{gregorian_year}")
    if choice == "2":
        print(f"民國年份：{roc_year}")
    print(f"對應的十二生肖為：{animal}")
    print("-" * 30)

if __name__ == "__main__":
    get_zodiac()

４. 在數學上0.1+0.2=0.3，可是下面的Python程式的答案好像不是0.3。怎麼解決？
```python
print(0.1+0.2)
```
